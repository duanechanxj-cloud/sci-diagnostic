# 04 — Generate Diagnostic

Select Approved items and assemble a balanced diagnostic across LO_IDs.


## Find eligible questions

Only Approved questions are eligible.


```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)

from src.question_bank import load_question_bank
from src.diagnostic import eligible_questions

bank = load_question_bank()
pool = eligible_questions(
    bank,
    level="P3",
    topic_codes=["P3-INT-MAG"],
    diagnostic_use="Topic",
)
print("Eligible:", len(pool))
```

## Generate

The round-robin selector spreads questions across available LO_IDs and favours lower-use items.


```python
from src.diagnostic import select_balanced_questions, coverage_table

selected = select_balanced_questions(
    pool,
    target_count=min(12, len(pool)),
    seed=2026,
)
display(selected[["Question_ID","LO_ID","Question_Text"]])
display(coverage_table(selected))
```
